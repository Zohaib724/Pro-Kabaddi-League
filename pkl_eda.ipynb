# **Statistical Test**

Checking average match total is the same in S1–4 and S5–10

```python
matches[matches["season"] <= 4]["total_points"].mean()
```

```python
matches[matches["season"] >= 5]["total_points"].mean()
```

```python
from scipy.stats import ttest_ind

ttest_ind(
    matches[matches["season"] <= 4]["total_points"],
    matches[matches["season"] >= 5]["total_points"],
    equal_var=False
)
```

Reject the null hypothesis. The average match total differs significantly between the two periods, with higher scoring in Seasons 5–10.

The p-value is far below 0.05, providing strong evidence against equal averages.

```python
matches.columns
```

Are match totals normally distributed?

```python
from scipy.stats import shapiro

shapiro(matches["total_points"])
```

The null hypothesis states that match totals follow a normal distribution. Since the p-value is less than 0.05, reject the null hypothesis

```python
matches["total_points"].skew()
```

Match totals depart significantly from a normal distribution, and the positive skewness indicates a mild right-skew.

```python
from scipy.stats import probplot

probplot(matches["total_points"], dist="norm", plot=plt)
plt.title("Q-Q Plot of Match Total Points")
plt.show()
```

Match totals are approximately normal in the middle but depart from normality in the tails. This agrees with the Shapiro–Wilk test and the positive skewness of 0.4423.
